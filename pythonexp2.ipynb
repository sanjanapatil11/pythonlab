```python
def check_a_before_b(s: str) -> bool:
    
    seen_b = False
    
    for char in s:
        if char == 'b':
            seen_b = True
        elif char == 'a' and seen_b:
           
            return False
            
    return True
s = "aabb"
print(check_a_before_b(s)) 

s = "abab"
print(check_a_before_b(s)) 

s = "bb"
print(check_a_before_b(s ))

```
